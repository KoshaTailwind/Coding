```python
set1 = {1, 2, 3}
set2 = {3, 4, 5}
print(set1 | set2) #объединение
print(set1 & set2) #пересечение
print(set1 - set2) #Разность
print(set1 ^ set2) #Симметрическая разнность
```

```python
print(set1.union(set2)) #объединение (union)
print(set1.intersection(set2)) #пересечение (intersetion)
print(set1.difference(set2)) #разность (diffference)
print(set1.symmetric_difference(set2)) #симметрическая разность (symmetric_difference)
```

LIST COMPREHENSION

```python
doubles = []
for x in range(10):
    doubles.append(x**2)
doubles
```

```python
doubles = [x**2 for x in range(10)]
doubles
```

```python
doubles = []
for x in range(10):
    if x %2 == 0:
        doubles.append(x**2)
doubles
```

```python
[x**2 for x in range(10) if x % 2 == 0]
```

```python
difficult = []
for x in range(3):
    for y in range(3):
        difficult.append((x, y))
difficult
```

```python
[(x, y) for x in range(3) for y in range(3)]
```

```python
doubles = {}
for x in range(5):
    doubles[x] = x**2
doubles
```

```python
{x: x**2 for x in range(5)}
```

COLLECTIONS

```python
from collections import Counter, defaultdict
```

```python
#Counter
fruits = ['apple', 'banana', 'apple', 'cherry']
fruit_count = Counter(fruits)
print(fruit_count)
```

```python
#defaultdict
colors = [('yellow', 1), ('blue', 2), ('yellow', 3), ('blue', 4), ('red', 5)]
default = defaultdict(list)
for key, value in colors:
    default[key].append(value)
default['banana']
dict(default.items())
```

```python
cats = 'catcatcat'
default = defaultdict(int)
for key in cats:
    default[key] += 1
default
```
