```python
import pandas as pd

df = pd.read_csv('previous_application.csv')
```

Задание 1. Загрузить источник данных – previous_application.csv, посчитать среднее значение платежа по всем кредитам для клиента с идентификатором 427000.

Ответ округлить до второго знака после запятой.

```python
# Фильтруем данные для клиента с ID 427000 и считаем среднее значение платежа
result = df[df['SK_ID_CURR'] == 427000]['AMT_ANNUITY'].mean()

# Округляем до двух знаков после запятой
rounded_result = round(result, 2)

print(f"Средний платеж для клиента 427000: {rounded_result}")
```

Задание 2. Загрузить источник данных previous_application.csv, вычислить количество активных одобренных кредитов.

```python
# Фильтруем одобренные и активные кредиты
approved_active_loans = df[
    (df['NAME_CONTRACT_STATUS'] == 'Approved') & 
    (df['DAYS_LAST_DUE'] == 365243)
]

# Считаем количество
count = len(approved_active_loans)

print(f"Количество активных одобренных кредитов: {count}")
```

Задание 3. Вычислить отношение суммарных платежей по текущим активным кредитам к зарплате. Ответ – значение для клиента с номером заявки 456254.
Скачать файл можно здесь.

Ответ округлить до 4-го знака после запятой.

```python
import pandas as pd

# Загружаем оба файла данных
prev_df = pd.read_csv('previous_application.csv')
app_df = pd.read_csv('application_train.csv')

# Фильтруем активные одобренные кредиты в предыдущих заявках
active_approved = prev_df[
    (prev_df['NAME_CONTRACT_STATUS'] == 'Approved') & 
    (prev_df['DAYS_LAST_DUE'] == 365243)
]

# Группируем по клиентам и считаем суммарный платеж
total_payments = active_approved.groupby('SK_ID_CURR')['AMT_ANNUITY'].sum().reset_index()
total_payments.columns = ['SK_ID_CURR', 'TOTAL_ACTIVE_PAYMENTS']

# Объединяем с основными данными
merged_df = pd.merge(app_df, total_payments, on='SK_ID_CURR', how='left')

# Заполняем пропущенные значения нулями (если нет активных кредитов)
merged_df['TOTAL_ACTIVE_PAYMENTS'] = merged_df['TOTAL_ACTIVE_PAYMENTS'].fillna(0)

# Вычисляем отношение суммарных платежей к зарплате
merged_df['PAYMENT_TO_INCOME_RATIO'] = merged_df['TOTAL_ACTIVE_PAYMENTS'] / merged_df['AMT_INCOME_TOTAL']

# Находим значение для клиента с ID 456254 и округляем
result = round(merged_df[merged_df['SK_ID_CURR'] == 456254]['PAYMENT_TO_INCOME_RATIO'].values[0], 4)

print(f"Отношение платежей к зарплате для клиента 456254: {result}")
```
