```python
#Сохраним IP key в переменную pixbay.com

API_KEY = "50761044-0847e7251d2ac899268376803"
```

```python
#Запишем в переменную наш 'Поисковой запрос' для изображения
image_to_find = 'kittten'

#Формируем url (адрес веб-сайта), по которому можно обращаться для получения списка изображений, 
#соответствующих поисковому запросу
url = f"https://pixabay.com/api/?key={API_KEY}&q={image_to_find}&image_type=photo"
print(url)
```

```python
#импортируем библиотеку urlib.request
#библиотека urlib включает в себя большое количество вспомогательных компонентов для работы с интернетом
#для удобства эти компоненты разделены в подбиблиотеки внутри библиотеки urlib
#здесь мы импортируем только нужную подбиблиотеку urlib.request - для выполнения HTTP запроса 
import urllib.request

#выполняем HTTP запрос к API pixbay
req = urllib.request.Request(url, headers={'User-Agent':'Mozilla/5.0'})
print(req)
```

```python
response = urllib.request.urlopen(req)
print(response)
```

```python
#Преобразовать HTTP ответ в строку
#Важно: функцию .read() можно позвать только 1 раз - второй раз читать уже будет нечего
#чтобы прочитать еще раз - нужно заново сделать HTTP запрос
data_string = response.read().decode()
```

```python
#напечатаем, что же получилось - выглядит не оч читаемо
print(f'type of data is: {type(data_string)}')
print(data_string)
```

```python
#на самом деле ответ от Pixabay API приходит в формате json
#используем библиотеку json

import json
data = json.loads(data_string)
```

```python
#напечатаем чт получилось теперь это уже питоновский словарь иии с ним можно обращаться как со словарем
#например доставать зннаения по ключам. все еще не очень читабельно

print(data)
```

```python
#обратимся как со словарем - достанем список ключей
print(data.keys())
```

```python
#мы уже знаем библиотеку pprint! используем её
from pprint import pprint
pprint(data)
```

```python
#импортируем библиотеку random
#эта библиотека облегчает работу со случайными числами
#в частности помогает сгенерировать случайное число с нужными критериями
#или выбрать случайный элемент из списка

import random

#выбираем случайное изображение из списка
random_hit = random.choice(data['hits'])
pprint(random_hit)
```

```python
#достанем из этой большой структуры сам url адрес изображения 
image_url = random_hit['largeImageURL']
print(image_url)
```

```python
#скачиваем само изображение
req = urllib.request.Request(image_url, headers={'User-Agent':'Mozilla/5.0'})
image_data = urllib.reques.urlopen(req).read()
```

```python
#посмотрим что там внутри image_data
#там бинарное представление изображения в jpg формате
print(image_data)
```

```python
#используем библиотеку IPython.display чтобыпоказать изображение прямо в jupyter notebook
from IPython.display import Image
Image(image_data, width=500)
```
