## Описание колонок

`Rank` - Общий рейтинг продаж

`Name` - Название игры

`Platform` - Платформа, на которой выпущена игра (например, PC, PS4 и т.д.)

`Year` - Год выпуска игры

`Genre` - Жанр игры

`Publisher` - Издатель игры

`NA_Sales` - Продажи в Северной Америке (в миллионах)

`EU_Sales` - Продажи в Европе (в миллионах)

`JP_Sales` - Продажи в Японии (в миллионах)

`Other_Sales` - Продажи в остальном мире (в миллионах)

`Global_Sales` - Общемировые продажи.

```python

import pandas as pd

# загрузите датасет
df = pd.read_csv('vgsales.csv')  # ваш код здесь

# после того датасет корректно загрузился, можно посмотреть на данные глазами
df.head()

df_clean = df[['Rank', 'Name', 'Platform', 'Year', 'Genre', 'Publisher', 'NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales', 'Global_Sales']]
df_clean = df_clean.copy()
```

### Дальше идут вопросы по датасету, на которые нужно найти ответ используя функционал библиотеки Pandas.

вопрос 1: Найдите количество уникальных игр в датасете.

```python
# ваш код здесь
nuniq_games = df_clean['Name'].nunique()
print(f'Количество уникальных игр: {nuniq_games}')
```

вопрос 2: Какова общая величина продаж игр в Японии?

```python
# ваш код здесь
df_Japan_sales = df_clean['JP_Sales'].sum()
print(f'Общая величина продаж игр в Японии: {df_Japan_sales}')
```

вопрос 3: Какие три жанра игр самые популярные в Северной Америке (по количеству продаж)? 

```python
# ваш код здесь
df_USA_popular = df_clean.groupby('Genre')['NA_Sales'].sum()
top3_sales = df_USA_popular.sort_values(ascending=False).head(3)

print('Три самых популрных жанра по продажам в Северной Америке:')
for Genre in top3_sales.index:
    print(Genre)
```

вопрос 4: Какая платформа была самой популярной (по количеству выпущенных игр) в 2000 году? А в 2015?

```python
# ваш код здесь
df_year_00 = df_clean[df_clean['Year'] == 2000]
games_platform_00 = df_year_00['Platform'].value_counts()

most_popular_plat00 = games_platform_00.idxmax()
most_popular_count00 = games_platform_00.max()

print(f'Самая популярная платформа в 2000 году: {most_popular_plat00} с {most_popular_count00} играми.')


df_year_15 = df_clean[df_clean['Year'] == 2015]
games_platform_15 = df_year_15['Platform'].value_counts()

most_popular_plat15 = games_platform_15.idxmax()
most_popular_count15 = games_platform_15.max()

print(f'Самая популярная платформа в 2015 году: {most_popular_plat15} с {most_popular_count15} играми.')
```

вопрос 5: Какой издатель выпустил больше всего игр в период 2012-2015 оба конца включительно?

```python
# ваш код здесь (ТУТ МНЕ БЫЛО ОЧ СЛОЖНО......)
df_clean = df_clean.dropna(subset=['Year'])
df_period = df_clean[(df_clean['Year'] >= 2012) & (df_clean['Year'] <= 2015)]
publisher_count = df_period['Publisher'].value_counts()
top_publisher = publisher_count.idxmax()
top_count = publisher_count.max()

print(f'Издатель, выпустивший больше всего игр в период 2012–2015 годов: {top_publisher} ({top_count} игр).')
```
